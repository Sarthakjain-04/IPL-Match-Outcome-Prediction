```python
deliveries.columns.tolist()
```

```python
deliveries['total_run'] = (
    deliveries['batsman_runs']+deliveries['extras']
)
```

```python
first_inning = deliveries[deliveries['inning']==1]
```

```python
first_inning.head()
```

```python
target_df = (
    first_inning.groupby('matchId')['total_run'].sum().reset_index()
)
```

```python
target_df['target'] = (
    target_df['total_run']+1
)
```

```python
target_df.head()
```

```python
match_df = matches_clean.merge(
    target_df[['matchId', 'target']],
    on = 'matchId'
)
```

```python
match_df.head()
```

```python
match_df.shape
```

```python
second_inning = deliveries[
    deliveries['inning']==2
].copy()
```

```python
second_inning.shape
```

```python
second_inning = second_inning.merge(
    match_df[['matchId','target']],
    on = 'matchId'
)
```

```python
second_inning.head()
```

```python
second_inning['current_score'] = (
    second_inning.groupby('matchId')['total_run'].cumsum()
)
```

```python
second_inning[['matchId','total_run','current_score']].head(20)
```

```python
second_inning['runs_left'] = (
    second_inning['target'] - second_inning['current_score']
)
```

```python
second_inning[['target','current_score','runs_left']].head()
```

```python
second_inning['balls_bowled'] = (
    second_inning['over']*6 + second_inning['ball']
)
```

```python
second_inning['balls_left'] = (
    120 - second_inning['balls_bowled']
)
```

```python
second_inning[
    ['over', 'ball', 'balls_bowled', 'balls_left']
    ].head(20)
```

```python
second_inning['player_dismissed'] = (
    second_inning['player_dismissed'].notna().astype(int)
)
```

```python
second_inning['wickets_fallen'] = (
    second_inning.groupby('matchId')['player_dismissed'].cumsum()
)
```

```python
second_inning[['player_dismissed','wickets_fallen']].head(50)
```

```python
second_inning['wickets_left'] = (
    10 - second_inning['wickets_fallen']
)
```

```python
second_inning['crr'] = (second_inning['current_score']*6)/(120 - second_inning['balls_left'])
```

```python
second_inning['rrr'] = (second_inning['runs_left']*6)/(second_inning['balls_left'])
```

```python
second_inning[
[
'target',
'current_score',
'runs_left',
'balls_left',
'wickets_left',
'crr',
'rrr'
]
].head(10)
```

```python
second_inning = second_inning.merge(
    match_df[
        ['matchId','winner','venue']
    ],
    on= 'matchId'
)
```

```python
second_inning.head()
```

```python
second_inning['result'] = np.where(
    second_inning['batting_team'] == second_inning['winner'],1,0
)
```

```python
second_inning[['batting_team', 'winner', 'result']].head(20)
```

```python
final_df = second_inning[
    [
        'batting_team',
        'bowling_team',
        'venue',
        'runs_left',
        'balls_left',
        'wickets_left',
        'target',
        'crr',
        'rrr',
        'result'
    ]
]
```

```python
final_df.head()
```

```python
final_df = final_df[final_df['balls_left']>0]
```

```python
final_df = final_df[final_df['runs_left']>=0]
```

```python
final_df.shape
```

```python
final_df['result'].value_counts()
```

```python
final_df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
x = final_df.drop('result',axis = 1)
y = final_df['result']
```

```python
x.head()
```

```python
y.head()
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size = 0.2,
    random_state = 42
)
```

```python
print(x_train.shape)
print(x_test.shape)
```

```python
trf = ColumnTransformer(
    [
        (
            'trf',OneHotEncoder(handle_unknown='ignore'),
            ['batting_team','bowling_team','venue']
        )
    ],
    remainder = 'passthrough'
)
```

```python
pipe = Pipeline(
    [
        ('step1',trf),
        ('step2',LogisticRegression(
            max_iter=3000
        ))
    ]
)
```

```python
pipe.fit(
    x_train,
    y_train
)
```

```python
y_pred = pipe.predict(x_test)
```

```python
accuracy_score(
    y_test,
    y_pred
)
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(
    y_test,
    y_pred
)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_pipe = Pipeline(
    [
        ('step1',trf),
        ('step2',RandomForestClassifier(
            n_estimators = 200,
            max_depth = 15,
            random_state = 42,
            n_jobs = -1
        ))
    ]
)
```

```python
rf_pipe.fit(x_train, y_train)
```

```python
rf_pred = rf_pipe.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
rf_accuracy = accuracy_score(
    y_test,
    rf_pred
)

print("Random Forest Accuracy:", rf_accuracy)
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(
    y_test,
    rf_pred
)
```

```python
from xgboost import XGBClassifier
```

```python
xgb_pipe = Pipeline([
    ('step1',trf),
    ('step2',XGBClassifier(
        n_estimators=300,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        eval_metric='logloss'))
])
```

```python
xgb_pipe.fit(x_train,y_train)
```

```python
xgb_pred = xgb_pipe.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score

xgb_accuracy = accuracy_score(
    y_test,
    xgb_pred
)

print("XGBoost Accuracy:", xgb_accuracy)
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(
    y_test,
    xgb_pred
)
```
